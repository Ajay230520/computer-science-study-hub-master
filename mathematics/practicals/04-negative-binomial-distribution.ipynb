# Practical 4 — Negative Binomial Distribution

**Aim:** Plot the PMF for the trial number on which the player's fifth win occurs, for red, black, and green.

**Scenario supplied:** 18 red, 10 black, 2 green slots; 30 equally likely slots total. This is a probability-model exercise, not gambling advice.

For the r-th success occurring on trial n: P(N=n)=C(n−1,r−1)p^r(1−p)^(n−r), n≥r. Here r=5 and n=1..20.

```python
import math
import numpy as np
import matplotlib.pyplot as plt

slots = {"Red": 18, "Black": 10, "Green": 2}
total_slots = sum(slots.values())
r = 5
n = np.arange(1, 21)
probabilities = {colour: count / total_slots for colour, count in slots.items()}

def negative_binomial_rth_success_pmf(n_trials, r, p):
    if n_trials < r:
        return 0.0
    return math.comb(n_trials - 1, r - 1) * (p ** r) * ((1 - p) ** (n_trials - r))

for colour, p in probabilities.items():
    print(f"{colour}: p={p:.6f}; mean trial for 5 wins = {r/p:.3f}")
```

```python
plt.figure(figsize=(10, 5))
for colour, p in probabilities.items():
    pmf = [negative_binomial_rth_success_pmf(int(k), r, p) for k in n]
    plt.plot(n, pmf, marker="o", label=f"{colour} (p={p:.3f})")
plt.xlabel("Trial n on which the 5th win occurs")
plt.ylabel("P(N = n)")
plt.title("Negative binomial PMF: fifth win by colour (n=1–20)")
plt.xticks(n)
plt.grid(alpha=0.3)
plt.legend()
plt.show()

print("Probability the 5th win occurs by trial 20:")
for colour, p in probabilities.items():
    cumulative = sum(negative_binomial_rth_success_pmf(int(k), r, p) for k in n)
    print(f"{colour}: {cumulative:.4%}")
```

## Observations

- The fifth win cannot occur before trial 5, so the PMF is zero for n<5.
- Higher success probability shifts the mass toward earlier trials. Red is expected to reach five wins sooner than black, and black sooner than green.
- The sum plotted over n=1..20 can be less than 1 because the fifth win may occur after trial 20.
- Assumptions: independent spins and constant success probability.

**Result:** Negative binomial PMFs were generated for the fifth win in each case.
