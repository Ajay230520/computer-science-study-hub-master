# Practical 5 — Binomial and Poisson Distributions

**Aim:** Generate binomial and Poisson distributions for the two scenarios in the supplied practical and calculate P(X=2) for each.

**Important source-data note:** The provided practical states Case 1 as 250 in 2500 (p=0.1) and Case 2 as 1 in 2500 (p=0.0004). These values differ by a factor of 250. The notebook preserves the supplied values as written; verify Case 1 with the instructor before treating it as a real-world prevalence estimate. This is an academic probability exercise, not medical guidance.

For X~Binomial(n,p), P(X=x)=C(n,x)p^x(1-p)^(n-x). The Poisson approximation uses λ=np and P(X=x)=e^(−λ)λ^x/x!.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, poisson

n = 1000
cases = {
    "Case 1: 250 / 2500": 250 / 2500,
    "Case 2: 1 / 2500": 1 / 2500,
}

results = []
for name, p in cases.items():
    lam = n * p
    exact_binom = binom.pmf(2, n, p)
    approx_poisson = poisson.pmf(2, lam)
    results.append((name, p, lam, exact_binom, approx_poisson))
    print(f"{name}\nn={n}, p={p:g}, q={1-p:g}, lambda=np={lam:g}")
    print(f"P(X=2), Binomial = {exact_binom:.12g}")
    print(f"P(X=2), Poisson  = {approx_poisson:.12g}\n")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
for ax, (name, p) in zip(axes, cases.items()):
    lam = n * p
    if p > 0.01:
        x = np.arange(max(0, int(lam - 4*np.sqrt(lam))), int(lam + 4*np.sqrt(lam)) + 1)
    else:
        x = np.arange(0, 11)
    ax.plot(x, binom.pmf(x, n, p), marker=".", label="Binomial")
    ax.plot(x, poisson.pmf(x, lam), marker=".", label="Poisson")
    ax.set_title(name)
    ax.set_xlabel("Number of cases (x)")
    ax.set_ylabel("P(X=x)")
    ax.grid(alpha=0.3)
    ax.legend()
plt.tight_layout()
plt.show()

print("Comparison summary")
for name, p, lam, pb, pp in results:
    print(f"{name}: lambda={lam:g}; Binomial P(X=2)={pb:.8g}; Poisson P(X=2)={pp:.8g}; absolute difference={abs(pb-pp):.3g}")
```

## Observations

- Case 1 has expected count np=100; exactly two cases is extraordinarily unlikely under the stated p=0.1 model.
- Case 2 has λ=np=0.4; the Poisson approximation closely matches the binomial result because n is large and p is small.
- The Poisson approximation is generally strongest for large n and small p with λ=np kept moderate; it is not equally accurate in every parameter regime.
- A distribution plot only shows the x-range selected; omitted tail probability remains outside the displayed range.

**Result:** Both distributions and P(X=2) were calculated for each supplied scenario.
