# Practical 1 — Classical and Empirical Probability

**Aim:** Implement classical and empirical probability, plot them, and discuss what happens as the number of trials increases.

**Environment:** Python 3, NumPy, Matplotlib. No paid service or external dataset required. The experiment uses a fair coin and a fixed random seed for reproducibility.

**Model:** Classical probability of heads for a fair coin is 1/2. Empirical probability is observed heads divided by the number of trials.

```python
import numpy as np
import matplotlib.pyplot as plt

SEED = 42
rng = np.random.default_rng(SEED)
TRIALS = 10_000
flips = rng.integers(0, 2, size=TRIALS)  # 1=heads, 0=tails
classical_p = 0.5
cumulative_empirical = np.cumsum(flips) / np.arange(1, TRIALS + 1)
print(f"Classical P(heads): {classical_p:.3f}")
print(f"Empirical P(heads) after {TRIALS:,} trials: {cumulative_empirical[-1]:.4f}")
```

```python
checkpoints = np.array([10, 50, 100, 500, 1_000, 5_000, 10_000])
print("Trials\tEmpirical P(heads)")
for n in checkpoints:
    print(f"{n}\t{cumulative_empirical[n-1]:.4f}")

plt.figure(figsize=(10, 5))
plt.plot(np.arange(1, TRIALS + 1), cumulative_empirical, label="Empirical probability")
plt.axhline(classical_p, linestyle="--", label="Classical probability = 0.5")
plt.xlabel("Number of coin flips")
plt.ylabel("Probability of heads")
plt.title("Classical vs. cumulative empirical probability")
plt.ylim(0, 1)
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

## Observations and conclusion

1. Empirical probability fluctuates strongly for small trial counts.
2. As trials accumulate, the cumulative estimate will often settle closer to 0.5, though it need not move monotonically or equal 0.5 exactly.
3. Classical probability is a model based on equally likely outcomes; empirical probability is estimated from observations.
4. A reproducible seed makes this run repeatable; a different seed may give a different trajectory.

**Result:** Classical and empirical probability were implemented and compared graphically.
