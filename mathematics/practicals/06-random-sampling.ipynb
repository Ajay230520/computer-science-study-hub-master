# Practical 6 — Random Sampling

**Aim:** Generate a population of 1,000 random integers, calculate population mean/variance/standard deviation, draw samples of sizes 10, 50, 100, 200, and 400, calculate the same statistics, and compare them.

**Reproducibility:** A fixed seed is used. Integers are sampled from a clearly documented range [1, 100]. Population variance uses ddof=0. Both population-style variance (ddof=0) and unbiased sample variance (ddof=1) are shown for sample subsets so the convention is explicit.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEED = 2026
rng = np.random.default_rng(SEED)
POPULATION_SIZE = 1000
LOW, HIGH_EXCLUSIVE = 1, 101
population = rng.integers(LOW, HIGH_EXCLUSIVE, size=POPULATION_SIZE)

pop_mean = float(np.mean(population))
pop_var = float(np.var(population, ddof=0))
pop_std = float(np.std(population, ddof=0))
print(f"Population size: {len(population)}; integer range: [{LOW}, {HIGH_EXCLUSIVE-1}]")
print(f"Population mean={pop_mean:.4f}, variance(ddof=0)={pop_var:.4f}, std(ddof=0)={pop_std:.4f}")
```

```python
sample_sizes = [10, 50, 100, 200, 400]
rows = []
for size in sample_sizes:
    sample = rng.choice(population, size=size, replace=False)
    rows.append({
        "sample_size": size,
        "mean": np.mean(sample),
        "variance_ddof0": np.var(sample, ddof=0),
        "variance_ddof1": np.var(sample, ddof=1),
        "std_ddof0": np.std(sample, ddof=0),
        "std_ddof1": np.std(sample, ddof=1),
    })
results = pd.DataFrame(rows)
results.round(4)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
metrics = [("mean", pop_mean, "Mean"), ("variance_ddof0", pop_var, "Variance (ddof=0)"), ("std_ddof0", pop_std, "Standard deviation (ddof=0)")]
for ax, (column, population_value, label) in zip(axes, metrics):
    ax.plot(results["sample_size"], results[column], marker="o", label="Sample statistic")
    ax.axhline(population_value, linestyle="--", label="Population statistic")
    ax.set_xlabel("Sample size")
    ax.set_ylabel(label)
    ax.set_title(label)
    ax.grid(alpha=0.3)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

print("Absolute differences from population statistics:")
for _, row in results.iterrows():
    print(f"n={int(row.sample_size):3d}: |mean diff|={abs(row['mean']-pop_mean):.4f}, |variance diff|={abs(row['variance_ddof0']-pop_var):.4f}, |std diff|={abs(row['std_ddof0']-pop_std):.4f}")
```

## Observations and conclusion

- Sample statistics vary with the random sample; increasing sample size often improves representativeness, but one particular sequence need not get closer at every step.
- The sample mean is an estimator of the population mean. Sample variance with ddof=1 corrects the usual finite-sample bias under standard random-sampling assumptions.
- Samples are drawn without replacement from the generated finite population. Each sample size uses a fresh draw; results depend on the fixed seed and should not be generalized from one run alone.
- Variance units are squared; standard deviation is in the original units.

**Result:** Population and sample statistics were computed and compared graphically.
