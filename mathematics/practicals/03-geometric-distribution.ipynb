# Practical 3 — Geometric Distribution

**Aim:** Plot the PMF for the trial on which the first win occurs when repeatedly betting on red, black, or green on the roulette wheel described in the practical.

**Source scenario:** 18 red, 10 black, 2 green slots; all 30 slots are treated as equally likely. This is a mathematical exercise, not gambling advice.

For success probability p, the geometric PMF is P(X=k)=(1-p)^(k-1)p for k=1,2,... .

```python
import numpy as np
import matplotlib.pyplot as plt

slots = {"Red": 18, "Black": 10, "Green": 2}
total_slots = sum(slots.values())
n = np.arange(1, 21)
probabilities = {colour: count / total_slots for colour, count in slots.items()}
print("Win probabilities:")
for colour, p in probabilities.items():
    print(f"{colour}: {p:.6f}")
```

```python
plt.figure(figsize=(10, 5))
for colour, p in probabilities.items():
    pmf = (1 - p) ** (n - 1) * p
    plt.plot(n, pmf, marker="o", label=f"{colour} (p={p:.3f})")
plt.xlabel("Trial n of the first win")
plt.ylabel("P(X = n)")
plt.title("Geometric PMF: first win by colour (trials 1–20)")
plt.xticks(n)
plt.grid(alpha=0.3)
plt.legend()
plt.show()

print("Probability first win occurs within 20 trials:")
for colour, p in probabilities.items():
    print(f"{colour}: {1 - (1-p)**20:.4%}")
```

## Observations

- Red has the largest per-spin win probability (18/30), so its PMF is highest at early trials and decays fastest.
- Black has probability 10/30; green has probability 2/30, so green's first win is less likely on any particular spin and its PMF decays more slowly.
- The plotted PMF covers n=1..20; probability mass beyond 20 is not shown.
- The model assumes independent spins and fixed probabilities.

**Result:** Geometric PMFs were generated for all three outcomes.
