# MFCS Practical 2 — Image Analysis and Histograms

**Subject classification:** MFCS practical list, as requested. The task itself uses image-analysis operations.
**Aim:** Read two images, analyse them, normalize pixel values, calculate min/max/mean/standard deviation/mode, plot separate histograms, and combine the histograms.

**Free assets:** `skimage.data.camera()` and `skimage.data.coins()` are sample images distributed with the open-source scikit-image package. They are loaded locally from the installed package; no paid API, login, or external download is required. Check scikit-image's image-data documentation and licensing before redistributing the raw image files independently.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from skimage import data

# Built-in, locally available sample assets from scikit-image.
image_a = data.camera()  # grayscale uint8 image
image_b = data.coins()   # grayscale uint8 image

print("Image A (camera):", image_a.shape, image_a.dtype)
print("Image B (coins):", image_b.shape, image_b.dtype)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].imshow(image_a, cmap="gray", vmin=0, vmax=255)
axes[0].set_title("Image A — camera")
axes[1].imshow(image_b, cmap="gray", vmin=0, vmax=255)
axes[1].set_title("Image B — coins")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()
```

```python
def image_statistics(image):
    values, counts = np.unique(image, return_counts=True)
    mode_value = values[np.argmax(counts)]
    return {
        "min": float(np.min(image)),
        "max": float(np.max(image)),
        "mean": float(np.mean(image)),
        "std": float(np.std(image)),
        "mode": int(mode_value),
    }

def normalize_0_1(image):
    image = image.astype(np.float64)
    low, high = image.min(), image.max()
    if high == low:
        return np.zeros_like(image, dtype=np.float64)
    return (image - low) / (high - low)

for name, image in [("Camera", image_a), ("Coins", image_b)]:
    print(name, image_statistics(image))

norm_a = normalize_0_1(image_a)
norm_b = normalize_0_1(image_b)
print("Normalized ranges:", (norm_a.min(), norm_a.max()), (norm_b.min(), norm_b.max()))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].imshow(norm_a, cmap="gray", vmin=0, vmax=1)
axes[0, 0].set_title("Camera — min-max normalized")
axes[0, 1].imshow(norm_b, cmap="gray", vmin=0, vmax=1)
axes[0, 1].set_title("Coins — min-max normalized")
axes[1, 0].hist(image_a.ravel(), bins=256, range=(0, 256), alpha=0.8, label="Camera")
axes[1, 0].set_title("Image A histogram")
axes[1, 1].hist(image_b.ravel(), bins=256, range=(0, 256), alpha=0.8, label="Coins")
axes[1, 1].set_title("Image B histogram")
for ax in axes.ravel():
    ax.grid(alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.hist(image_a.ravel(), bins=256, range=(0, 256), alpha=0.55, label="Camera")
plt.hist(image_b.ravel(), bins=256, range=(0, 256), alpha=0.55, label="Coins")
plt.xlabel("Grayscale intensity (0=black, 255=white)")
plt.ylabel("Pixel count")
plt.title("Combined histogram of both images")
plt.legend()
plt.grid(alpha=0.2)
plt.show()

# Optional: compare normalized intensities on a common 0..1 scale.
plt.figure(figsize=(10, 4))
plt.hist(norm_a.ravel(), bins=100, range=(0, 1), alpha=0.55, label="Camera normalized")
plt.hist(norm_b.ravel(), bins=100, range=(0, 1), alpha=0.55, label="Coins normalized")
plt.xlabel("Normalized intensity")
plt.ylabel("Pixel count")
plt.title("Combined histogram after per-image min-max normalization")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

## Observations and conclusion

- Min-max normalization maps each image's minimum to 0 and maximum to 1. It does not necessarily make the two images have the same intensity distribution.
- The mean summarizes average brightness; standard deviation measures intensity spread. The mode is the most frequent pixel intensity; ties are resolved by choosing the smallest intensity returned by `np.unique`.
- A histogram shows the distribution of pixel intensities but discards spatial arrangement: two images can have similar histograms and very different structures.
- The combined raw histogram overlays pixel counts. Since image dimensions may differ, raw counts are affected by image size; use density-normalized histograms for distribution-shape comparisons.

**Result:** Two free sample images were loaded, analysed, normalized, and their individual and combined histograms were plotted.
